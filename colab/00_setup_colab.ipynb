# Come usare Google Colab per il laboratorio

## Cosa serve
- Un account Google (quello di Gmail va benissimo).
- Niente chiavi, niente installazioni. Tutto gira nel browser.

## Tempo necessario: 2 minuti

## Passo 1 — Apri Google Colab

Vai su [colab.research.google.com](https://colab.research.google.com) e accedi con il tuo account Google.

## Passo 2 — Apri i notebook del corso

Per ogni notebook:
1. Il docente ti dà il link → cliccalo, si apre in Colab.
2. In alto compare **"Copia in Drive"**: cliccalo. Così lavori sulla **tua** copia e le modifiche restano.
3. Esegui le celle dall'alto in basso con **Shift+Invio**.

La prima volta che esegui una cella, Colab chiede "Eseguire comunque?" → **Esegui comunque**.

## Passo 3 — Verifica che funziona

Esegui la cella qui sotto (clicca dentro e premi **Shift+Invio**). Deve stampare due righe e finire con "Tutto pronto".

```python
# Verifica: Python funziona e la rete del laboratorio lascia passare le chiamate HTTP
import sys
import requests

print("Python", sys.version.split()[0])

response = requests.get("https://jsonplaceholder.typicode.com/todos/1", timeout=10)
print("Chiamata HTTP di prova -> status code", response.status_code)

if response.status_code == 200:
    print("Tutto pronto.")
else:
    print("Qualcosa blocca la rete: chiama il docente.")
```

## Scorciatoie utili

| Tasto | Azione |
|-------|--------|
| `Shift+Invio` | Esegui la cella e passa alla successiva |
| `Ctrl+Invio` | Esegui la cella e resta lì |
| `Ctrl+M B` | Nuova cella sotto |
| `Ctrl+M D` | Cancella la cella |

## Se una cella si blocca

Menu **Runtime → Interrompi esecuzione**. Se il notebook è "impazzito": **Runtime → Riavvia sessione** e riesegui dall'alto.
